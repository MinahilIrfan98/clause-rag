### RAG (Retrieval Augmented Generation)


2 Step RAG

- Retrieval -- Embeddings (Query, Document(s))
- Generation -- LLM (Query + Context)


```python
# Generative Model or LLM --- Ollama, OpenAI
# Embedding Model -- Ollama, OpenAI

#
```

```python
from langchain.chat_models import init_chat_model # LLM
from langchain.embeddings import init_embeddings  # Embeddings
# from langchain_community.document_loaders import PyPDFLoader
from langchain_core.documents import Document # .txt, .md
from langchain_text_splitters import RecursiveCharacterTextSplitter # chunks
from langchain_core.vectorstores import InMemoryVectorStore # vs
```

```python
text = open(file="doc.txt", encoding="utf-8").read() # string object
```

```python
doc = Document(page_content=text, metadata={"source": "doc.txt"})
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=600, chunk_overlap=100)
```

```python
chunks = splitter.split_documents([doc])
```

```python
len(chunks)
```

```python
embedd = init_embeddings(model="nomic-embed-text:latest", provider="ollama")
```

```python
vs = InMemoryVectorStore.from_documents(
    documents=chunks,
    embedding=embedd
)
```
